we are going to create and measure a single qubit using Qiskit. The goal is to understand how a qubit is represented in a quantum circuit, how a quantum gate changes its state, and how measurement converts that quantum state into a classical result. We will also use Qiskit Aer to simulate the circuit, allowing the complete process to run without physical quantum hardware.

```python
# just editted
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
```

```python
circuit = QuantumCircuit(1, 1)

print(circuit)
```

```python
circuit.x(0)

circuit.draw("mpl")
```

```python
circuit.measure(0, 0)

circuit.draw("mpl")
```

```python
simulator = AerSimulator()
```

```python
result = simulator.run(
    circuit,
    shots=1000
).result()

counts = result.get_counts()

print(counts)
```

```python
plot_histogram(counts, color='orange')
```

```python
default_circuit = QuantumCircuit(1, 1)

default_circuit.measure(0, 0)

default_circuit.draw("mpl")
```

```python
default_result = simulator.run(
    default_circuit,
    shots=1000
).result()

default_counts = default_result.get_counts()

print(default_counts)
```

```python
plot_histogram(default_counts)
```

```python
plot_histogram(
    [default_counts, counts],
    legend=["Initial |0> State", "After X Gate"],
    color=['#1f77b4', 'orange']
)
```
