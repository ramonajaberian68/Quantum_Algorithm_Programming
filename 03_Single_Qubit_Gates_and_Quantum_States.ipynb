we are going to apply several single-qubit gates and visualise how each gate changes the quantum state. The goal is to move beyond simply reading measurement results and instead see how quantum gates transform a qubit on the Bloch sphere.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
```

```python
initial_circuit = QuantumCircuit(1)

initial_state = Statevector.from_instruction(initial_circuit)

print(initial_state)
```

```python
plot_bloch_multivector(initial_state)
```

```python
x_circuit = QuantumCircuit(1)

x_circuit.x(0)

x_circuit.draw("mpl")
```

```python
x_state = Statevector.from_instruction(x_circuit)

print(x_state)

plot_bloch_multivector(x_state)
```

```python
h_circuit = QuantumCircuit(1)

h_circuit.h(0)

h_circuit.draw("mpl")
```

```python
h_state = Statevector.from_instruction(h_circuit)

print(h_state)

plot_bloch_multivector(h_state)
```

```python
y_circuit = QuantumCircuit(1)

y_circuit.y(0)

y_circuit.draw("mpl")
```

```python
y_state = Statevector.from_instruction(y_circuit)

print(y_state)

plot_bloch_multivector(y_state)
```

```python
z_circuit = QuantumCircuit(1)

z_circuit.h(0)
z_circuit.z(0)

z_circuit.draw("mpl")
```

```python
z_state = Statevector.from_instruction(z_circuit)

print(z_state)

plot_bloch_multivector(z_state)
```

```python
s_circuit = QuantumCircuit(1)

s_circuit.h(0)
s_circuit.s(0)

s_circuit.draw("mpl")
```

```python
s_state = Statevector.from_instruction(s_circuit)

print(s_state)

plot_bloch_multivector(s_state)
```

```python
t_circuit = QuantumCircuit(1)

t_circuit.h(0)
t_circuit.t(0)

t_circuit.draw("mpl")
```

```python
t_state = Statevector.from_instruction(t_circuit)

print(t_state)

plot_bloch_multivector(t_state)
```

```python
states = {
    "Initial |0>": initial_state,
    "X Gate": x_state,
    "Y Gate": y_state,
    "H Gate": h_state,
    "H + Z": z_state,
    "H + S": s_state,
    "H + T": t_state
}

for name, state in states.items():
    print(f"{name}: {state.data}")
```
